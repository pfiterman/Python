```python
# Constants
price_apple = 0.5   # Cost per apple in dollars
price_banana = 0.25  # Cost per banana in dollars
price_orange = 0.4    # Cost per orange in dollars


# Input from the user
num_apples = int(input("Enter the number of apples: "))
num_bananas = int(input("Enter the number of bananas: "))
num_oranges = int(input("Enter the number of oranges: "))

# Calculate total cost
total_cost = (num_apples * price_apple) + \
             (num_bananas * price_banana) + \
             (num_oranges * price_orange)

# Display the result
print("Total cost of the items: $", total_cost)
```
