```python
import pandas as pd
```

```python
df = pd.DataFrame()
```

```python
df["Stock"] = ["MSFT", "AAPL", "IBM", "ORCL"]
df["Price"] = [200, 300, 50, 100]
```

```python
df["Shares"] = [100, 200, 300, 400]
df["Amnt"] = df["Price"] * df["Shares"]
df
```

```python
condition = df["Price"] < 100
df[condition]
```

```python
condition_max = df["Amnt"]==df["Amnt"].max()
df[condition_max]
```
