```python
import pandas as pd
```

```python
df = pd.DataFrame()
```

```python
df["Food"] = ["Pizza","Steak","Salad","Hamburger"]
df["Price"] = [26.5, 30.9,14.9, 23.9]
```

```python
df["Qnty"] = [3, 5, 3,2]
```

```python
df["Amnt"] = df["Price"] * df["Qnty"]
```

```python
condition_20 = df["Price"]<20
df[condition_20]
```

```python
df["Amnt"].std()
```

```python
max_value = df["Amnt"].max()
max_value_condition = df["Amnt"] == max_value
df[max_value_condition]
```

```python
import matplotlib.pyplot as plt
```

```python
x = df["Amnt"]
labels = df["Food"]
```

```python
colors = ["#DFFF00","#FFBF00","#FF7F50","#DE3163","#9FE2BF"]
e = [0,0.2,0,0]
plt.pie(x, labels=labels,shadow=True,explode=e,colors=colors, autopct="%.1f%%");
plt.title("Pie Chart", color="blue");
```
